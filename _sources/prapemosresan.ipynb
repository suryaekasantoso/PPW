# Pra-Pemrosesan Teks

Preprocessing teks adalah tahapan awal yang sangat penting dalam pengolahan data berbasis teks. Data teks yang diperoleh dari berbagai sumber, seperti abstrak penelitian, artikel berita, hingga media sosial, biasanya masih berantakan dan mengandung banyak unsur yang tidak relevan. Unsur tersebut bisa berupa huruf kapital yang tidak konsisten, tanda baca, angka, simbol, kata tidak baku, bahkan kesalahan penulisan. Kondisi ini dapat menurunkan kualitas analisis karena komputer akan kesulitan membedakan mana informasi yang penting dan mana yang sekadar noise. Oleh karena itu, preprocessing diperlukan agar data teks menjadi lebih bersih, konsisten, dan siap diproses lebih lanjut.

Tujuan utama dari preprocessing teks bukan hanya sekadar membersihkan data, tetapi juga untuk memastikan bahwa teks dapat diproses dengan lebih efektif dan menghasilkan informasi yang bermanfaat. Secara lebih rinci, tujuan tersebut antara lain:

1. Meningkatkan kualitas data

    Meningkatkan kualitas berarti membersihkan elemen yang mengganggu sehingga teks lebih representatif terhadap isi sebenarnya. Elemen ini meliputi noise teknis (HTML tag, karakter non-UTF), metadata yang tidak relevan (nama pembimbing, alamat email), tanda baca berlebih, hasil OCR yang keliru, dan baris kosong.
2. Menyeragamkan representasi teks

    Membuat variasi penulisan yang sama diperlakukan identik supaya sistem nggak menganggapnya berbeda. Ini termasuk case folding, normalisasi singkatan/gaul, perluasan singkatan (BEI → Bursa Efek Indonesia jika perlu), dan normalisasi whitespace/punctuation.
3. Mengurangi kompleksitas data

    Mengurangi jumlah fitur/token yang harus diproses tanpa menghilangkan informasi penting. tujuannya menghemat memori, mempercepat training, dan mengurangi kebisingan statistik.
4. Memperkuat makna dalam analisis

    Langkah seperti stemming/lemmatization dan pembentukan n-gram menyatukan variasi morfologis sehingga makna inti lebih tersorot. Misal kata berkaitan (berinvestasi, investasi, investor) dapat distandarkan agar frekuensi mencerminkan topik sebenarnya.

```python
def get_max_page(prodi_id):
    url = f"{BASE_URL}/{prodi_id}/1"
    r = requests.get(url)
    soup = BeautifulSoup(r.content, "html.parser")

    # Cari tombol last page
    last_page = soup.select_one('ol.pagination a:contains("»")')
    if last_page and "href" in last_page.attrs:
        href = last_page["href"]
        max_page = int(href.split("/")[-1])
        return max_page

    # fallback kalau pagination tidak ada
    return 1
```

```python
def print_progress(prodi_id, prodi, current_page, total_pages):
    percent = (current_page / total_pages) * 100
    bar_length = 20
    filled_length = int(bar_length * current_page // total_pages)
    bar = '█' * filled_length + '-' * (bar_length - filled_length)
    sys.stdout.write(f'\r[{prodi_id}] {prodi} - Page {current_page}/{total_pages} [{bar}] {percent:.2f}%')
    sys.stdout.flush()
    if current_page == total_pages:
        sys.stdout.write('\n')
```

## Crawling Data PTA (Prodi Manajemen)

Fungsi pta() yang digunakan untuk melakukan web crawling data skripsi/pta (Program Tugas Akhir) dari Prodi Manajemen. Prosesnya dimulai dengan menentukan daftar prodi yang akan diambil datanya, lalu menghitung jumlah halaman yang tersedia untuk setiap prodi. Selanjutnya, fungsi akan melakukan scraping pada tiap halaman untuk mengumpulkan informasi penting seperti ID PTA, nama penulis, judul, abstrak (bahasa Indonesia dan Inggris), nama dosen pembimbing I dan II, serta nama prodi. Data tersebut diambil dari struktur HTML menggunakan BeautifulSoup, kemudian disimpan dalam bentuk dictionary dan dikonversi menjadi DataFrame pandas. Setelah semua data terkumpul, hasilnya diekspor ke file Excel (pta.xlsx). Selain itu, fungsi juga menampilkan progress crawling per halaman serta ringkasan akhir berupa total entri yang berhasil dikumpulkan dan estimasi waktu eksekusi program.

```python
def pta():
    start_time = time.time()

    data = {
        "id": [],
        "penulis": [],
        "judul": [],
        "abstrak_id": [],
        "abstrak_en": [],
        "pembimbing_pertama": [],
        "pembimbing_kedua": [],
        "prodi": []
    }

    # daftar prodi yang akan diproses
    prodi_list = [7]
    total_pages = 0
    max_pages_dict = {}

    # hitung total halaman (untuk tiap prodi yang dipilih)
    for i in prodi_list:
        max_page = get_max_page(i)
        max_pages_dict[i] = max_page
        total_pages += max_page

    # scraping data tiap prodi
    for i in prodi_list:
        max_page = max_pages_dict[i]
        for j in range(1, max_page + 1):
            url = f"{BASE_URL}/{i}/{j}"
            r = requests.get(url)
            soup = BeautifulSoup(r.content, "html.parser")
            jurnals = soup.select('li[data-cat="#luxury"]')

            isii = soup.select_one('div#begin')
            if not isii:
                continue
            prodi_full = isii.select_one('h2').text.strip()
            prodi = prodi_full.replace("Journal Jurusan ", "")

            for jurnal in jurnals:
                link_keluar = jurnal.select_one('a.gray.button')['href']

                # ambil ID dari link PTA
                id_match = re.search(r"/detail/(\d+)", link_keluar)
                pta_id = id_match.group(1) if id_match else None

                response = requests.get(link_keluar)
                soup1 = BeautifulSoup(response.content, "html.parser")
                isi = soup1.select_one('div#content_journal')

                judul = isi.select_one('a.title').text.strip()
                penulis = isi.select_one('span:contains("Penulis")').text.split(' : ')[1]
                pembimbing_pertama = isi.select_one('span:contains("Dosen Pembimbing I")').text.split(' : ')[1]
                pembimbing_kedua = isi.select_one('span:contains("Dosen Pembimbing II")').text.split(' :')[1]

                paragraf = isi.select('p[align="justify"]')
                abstrak_id = paragraf[0].get_text(strip=True) if len(paragraf) > 0 else "N/A"
                abstrak_en = paragraf[1].get_text(strip=True) if len(paragraf) > 1 else "N/A"

                data["id"].append(pta_id)
                data["penulis"].append(penulis)
                data["judul"].append(judul)
                data["abstrak_id"].append(abstrak_id)
                data["abstrak_en"].append(abstrak_en)
                data["pembimbing_pertama"].append(pembimbing_pertama)
                data["pembimbing_kedua"].append(pembimbing_kedua)
                data["prodi"].append(prodi)

            # update progress bar per prodi
            print_progress(i, prodi, j, max_page)

    # simpan ke Excel
    df = pd.DataFrame(data)
    df.to_excel("pta.xlsx", index=False, engine='openpyxl')

    # hitung durasi
    end_time = time.time()
    elapsed = int(end_time - start_time)
    jam, sisa = divmod(elapsed, 3600)
    menit, detik = divmod(sisa, 60)

    # summary
    print("\n✅ Seluruh data berhasil dikumpulkan!")
    print(f"📊 Total entri: {len(df)}")
    print(f"⏱️ Waktu eksekusi: {jam} jam {menit} menit {detik} detik")

    return None
```

### Pembersihan Teks

Tahap pembersihan teks merupakan proses awal dalam text preprocessing yang bertujuan untuk menghilangkan elemen-elemen yang tidak relevan sehingga data lebih bersih, seragam, dan siap dianalisis. Data teks mentah, seperti abstrak penelitian, umumnya masih mengandung berbagai noise seperti karakter khusus, tanda baca, angka, maupun spasi berlebih yang dapat mengganggu proses analisis lebih lanjut. Oleh karena itu, dilakukan pembersihan untuk memastikan teks berada dalam bentuk standar.

Pada penelitian ini, pembersihan teks dilakukan melalui beberapa langkah, antara lain:

- Menghapus karakter khusus seperti _x000D_ yang muncul akibat konversi file Excel.
- Menghilangkan line break (\n, \r) agar teks berada dalam satu baris.
- Mengubah seluruh huruf menjadi huruf kecil (case folding) untuk    menyeragamkan penulisan.
- Mengubah emoji menjadi teks deskriptif menggunakan fungsi emoji.demojize() agar tidak hilang makna.
- Menghapus angka dan tanda baca sehingga hanya tersisa huruf.
- Menghapus karakter non-alfabet dan menggantinya dengan spasi.
- Membersihkan elemen HTML yang mungkin terbawa saat proses scraping dengan bantuan BeautifulSoup.
- Merapikan spasi ganda agar teks lebih terstruktur dan mudah diproses.

Hasil pembersihan disimpan dalam kolom baru yaitu abstrak_id_clean, sehingga data asli tetap terjaga sementara data bersih dapat digunakan pada tahap analisis berikutnya, seperti tokenization, stopword removal, maupun stemming. Dengan adanya tahap ini, kualitas data teks meningkat dan analisis dapat dilakukan dengan lebih akurat serta konsisten.

```python
# Fungsi cleaning dengan emoji
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = text.replace('_x000D_', ' ')  # hapus karakter Excel khusus
    text = text.replace('\n', ' ').replace('\r', ' ')  # hapus line break
    text = text.lower()
    text = emoji.demojize(text)
    text = re.sub(r'\d+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\W', ' ', text)
    text = BeautifulSoup(text, "html.parser").get_text()
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Baca Excel
pta_df = pd.read_excel("pta.xlsx", engine='openpyxl', dtype=str).fillna("")

# Terapkan langsung ke kolom target
pta_df["abstrak_id_clean"] = pta_df["abstrak_id"].apply(clean_text)
```

### Tokenisasi

Tokenisasi merupakan tahap dalam text preprocessing yang berfungsi untuk memecah teks yang sudah dibersihkan menjadi potongan-potongan kata kecil yang disebut token. Proses ini sangat penting karena komputer tidak dapat langsung memahami teks dalam bentuk kalimat panjang, melainkan perlu memprosesnya dalam unit yang lebih sederhana. Dengan tokenisasi, analisis lebih lanjut seperti stopword removal, stemming, maupun perhitungan frekuensi kata dapat dilakukan dengan lebih mudah.

Pada penelitian ini, tokenisasi dilakukan menggunakan fungsi word_tokenize dari pustaka NLTK (Natural Language Toolkit). Fungsi ini memecah teks pada kolom abstrak_id_clean menjadi daftar kata, kemudian hasilnya disimpan pada kolom baru bernama abstrak_id_tokens. Dengan demikian, setiap abstrak penelitian yang semula berupa teks panjang berubah menjadi kumpulan kata-kata terpisah yang siap dianalisis.

```python
# Tokenisasi untuk PTA
pta_df["abstrak_id_tokens"] = pta_df["abstrak_id_clean"].apply(word_tokenize)
```

```python
print("\nPTA (abstrak_id_tokens):")
pta_df[["abstrak_id_clean", "abstrak_id_tokens"]].head(5)
```

### Penghapusan Kata Umum (Stopwords)

Stopword removal adalah proses menghapus kata-kata umum (stopwords) yang sering muncul dalam teks tetapi tidak memiliki makna penting dalam analisis. Kata-kata seperti yang, dan, atau, adalah, dengan, untuk, dan sejenisnya sering muncul dalam kalimat, namun tidak memberikan informasi spesifik terkait isi dokumen. Jika tidak dihapus, kata-kata tersebut justru dapat menambah noise dan menurunkan akurasi analisis teks.

Dalam penelitian ini, daftar stopwords bahasa Indonesia diambil dari pustaka NLTK. Prosesnya dilakukan dengan memfilter setiap token hasil tokenisasi, lalu hanya menyimpan kata-kata yang tidak termasuk dalam daftar stopwords. Hasil akhir disimpan pada kolom baru bernama abstrak_id_filtered. Dengan demikian, teks yang semula penuh dengan kata umum menjadi lebih ringkas dan fokus pada kata-kata bermakna.

```python
# Stopwords untuk bahasa Indonesia
stop_words_id = set(stopwords.words('indonesian'))

# Filter stopwords di PTA
pta_df["abstrak_id_filtered"] = pta_df["abstrak_id_tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words_id]
)
```

```python
print("\nPTA (abstrak_id_filtered):")
pta_df[["abstrak_id_tokens", "abstrak_id_filtered"]].head(5)
```

### Stemming

Stemming adalah proses mengubah kata-kata yang sudah melalui tahap tokenization dan stopword removal menjadi bentuk dasarnya (root word). Tujuannya adalah menyeragamkan berbagai bentuk kata yang berbeda namun memiliki makna yang sama. Misalnya, kata “berlari”, “lari”, dan “pelari” semuanya akan dikembalikan ke bentuk dasar yaitu “lari”. Dengan demikian, analisis menjadi lebih konsisten karena tidak ada perbedaan perlakuan antara variasi kata yang sebenarnya memiliki makna sama.

Dalam penelitian ini, proses stemming menggunakan Sastrawi, yaitu pustaka Python yang dirancang khusus untuk bahasa Indonesia. Dengan bantuan StemmerFactory, dibuat objek stemmer (indo_stemmer) yang kemudian diterapkan pada setiap token dalam kolom abstrak_id_filtered. Hasil stemming disimpan dalam kolom baru bernama abstrak_id_stemmed.

```python
factory = StemmerFactory()
indo_stemmer = factory.create_stemmer()

pta_df["abstrak_id_stemmed"] = pta_df["abstrak_id_filtered"].apply(
    lambda tokens: [indo_stemmer.stem(word) for word in tokens]
)
```

### Pemeriksaan Ejaan

Tahap pemeriksaan ejaan bertujuan untuk memperbaiki kesalahan penulisan kata dalam data teks, baik akibat typo, kesalahan input, maupun variasi penulisan yang tidak sesuai dengan kaidah bahasa Indonesia. Proses ini penting agar kata yang dianalisis benar-benar sesuai dengan bentuk bakunya dan tidak menimbulkan perhitungan ganda dalam analisis frekuensi kata maupun pemodelan teks.

Dalam penelitian ini digunakan library pyspellchecker dengan kamus bahasa Indonesia yang diambil dari file 00-indonesian-wordlist.lst. Kamus tersebut berisi daftar kosakata bahasa Indonesia yang sudah distandarisasi, sehingga setiap kata yang ditemukan di luar daftar akan diperiksa dan dikoreksi.

```python
# Inisialisasi SpellChecker kosong
spell = SpellChecker(language=None)

# Load kamus Indonesia dari file
with open("00-indonesian-wordlist.lst", "r", encoding="latin-1") as f:
    indo_words = [line.strip() for line in f.readlines()]

spell.word_frequency.load_words(indo_words)

# Fungsi untuk koreksi kata
def correct_word(word):
    corr = spell.correction(word)
    return corr if corr is not None else word

# Terapkan spellcheck ke setiap baris dengan progress bar
corrected_texts = []
for tokens in tqdm(pta_df["abstrak_id_expanded"], desc="Spellchecking", unit="row"):
    corrected = [correct_word(word) for word in tokens]
    corrected_texts.append(corrected)

pta_df["abstrak_id_spellchecked"] = corrected_texts
```

## Analisis Frekuensi Kata

Setelah melalui tahapan preprocessing (pembersihan, tokenisasi, penghapusan stopwords, stemming, ekspansi kontraksi, dan pemeriksaan ejaan), langkah berikutnya adalah melakukan analisis terhadap frekuensi kata yang muncul dalam kumpulan data. Analisis ini bertujuan untuk mengetahui kata-kata yang paling sering digunakan, sehingga dapat memberikan gambaran umum mengenai tema utama dan pola bahasa dalam teks.

```python
# Gabungkan semua token jadi satu list besar
all_tokens = [word for tokens in pta_df["abstrak_id_spellchecked"] for word in tokens]

# Hitung frekuensi kata
word_freq = Counter(all_tokens)

# Ambil 20 kata paling sering muncul
common_words = word_freq.most_common(20)

print("=== 20 Kata Terbanyak ===")
for word, freq in common_words:
    print(f"{word} : {freq}")

# Informasi tambahan
total_words = len(all_tokens)                       # jumlah total kata
unique_words = len(word_freq)                       # jumlah kata unik
avg_words_per_data = np.mean([len(tokens) for tokens in pta_df["abstrak_id_spellchecked"]])  # rata-rata jumlah kata per data

print("\n=== Statistik Tambahan ===")
print(f"Jumlah total kata       : {total_words}")
print(f"Jumlah kata unik        : {unique_words}")
print(f"Rata-rata kata per data : {avg_words_per_data:.2f} kata")
```
